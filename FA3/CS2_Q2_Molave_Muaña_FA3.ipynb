```python

menu = [ "Display Scores", "Search Score", "Compute Average Score ", "Update Score ", "Display Updated Scores", "Exit"]

scores = [50, 70, 88, 90, 100]

#Functions required for the program to function.

def display_scores():
  for i in range(len(scores)):
    print(f"Index {i}: {scores[i]}")


def search_score():
  target = int(input("Enter score to search: "))
  if target in scores:
    print(f"Found at index {scores.index(target)}")
  else:
    print("Score not found.")


def compute_average():
  print("Average Score:", sum(scores) / len(scores))


def update_score():
  index = int(input("Enter index to update (0-4): "))
  scores[index] = int(input("Enter new score: "))
  print("Updated!")

choice = ""

#Places the user in a while loop unless they input 6.
while choice != "6":

  #Index/Choices displayed once the user runs the program.

  print("\n=== Student Score Analyzer ===")
  print("1: Display Scores")
  print("2: Search Score")
  print("3: Compute Average Score")
  print("4: Update Score")
  print("5: Display Updated Scores")
  print("6: Exit")

  choice = input("\nEnter your choice: ")

#Conditions the choices the user must meet in order for them to function in the program.
  if choice == "1":
    display_scores()
  elif choice == "2":
    search_score()
  elif choice == "3":
    compute_average()
  elif choice == "4":
    update_score()
  elif choice == "5":
    display_scores()
  elif choice == "6":
    print("Goodbye!")
  else:
    print("Invalid choice.")

```
